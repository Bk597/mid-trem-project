```python
class Car:
    def __init__(self, brand, model):#dunder method
        self.brand = brand
        self.model = model

# Creating objects
car1 = Car("Toyota", "Corolla")
car2 = Car("Honda", "Civic")

print(car1.brand, car1.model)
print(car2.brand, car2.model)
```

```python
class Student:
    def __init__(self, name, age):
        self.name = name      # instance variable
        self.age = age        # instance variable

    def show_info(self):      # instance method
        print(f"Name: {self.name}, Age: {self.age}")

s1 = Student("Ali", 20)
s2 = Student("Sara", 22)

s1.show_info()
s2.show_info()
```

```python
class Company:
    company_name = "Tech Corp"   # class variable

    def __init__(self, emp_name):
        self.emp_name = emp_name   # instance variable

    @classmethod
    def show_company(cls):         # class method
        print("Company:", cls.company_name)

emp1 = Company("Ali")
emp2 = Company("Sara")

print(emp1.emp_name)
Company.show_company()
```

```python
class Animal:
    def sound(self):
        print("This is an animal sound.")

class Dog(Animal):   # inherits from Animal
    def sound(self):
        print("Bark!")

d = Dog()
d.sound()
```

```python
class Grandparent:
    def speak(self):
        print("I am the grandparent.")

class Parent(Grandparent):
    def speak(self):
        print("I am the parent.")

class Child(Parent):
    def speak(self):
        print("I am the child.")

c = Child()
c.speak()
```

```python
class Shape:
    def area(self):
        print("This is a generic shape.")

class Circle(Shape):
    def area(self):
        print("Area = πr²")

class Square(Shape):
    def area(self):
        print("Area = side²")

c = Circle()
s = Square()
c.area()
s.area()
```

```python
class A:
    def show(self):
        print("Class A")

class B:
    def show(self):
        print("Class B")

class C(A, B):   # multiple inheritance
    pass

c = C()
c.show()  # Which one will be called? Check MRO
print(C.mro())
```

```python
class Example:
    def __init__(self):
        self.public = "I am Public"
        self._protected = "I am Protected"
        self.__private = "I am Private"

    def show(self):
        print(self.public)
        print(self._protected)
        print(self.__private)

obj = Example()
obj.show()

print(obj.public)       # works
print(obj._protected)   # works but not recommended
#print(obj.__private)  # ERROR: private not accessible directly
```

```python
class University:
    def __init__(self, name):
        self.name = name

    class Department:
        def __init__(self, dept_name):
            self.dept_name = dept_name

uni = University("My University")
dept = University.Department("Computer Science")
print(uni.name, "-", dept.dept_name)

```

```python
# Association Example
class Author:
    def __init__(self, name):
        self.name = name

class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author

auth = Author("J.K. Rowling")
book = Book("Harry Potter", auth)
print(book.title, "by", book.author.name)
```

```python
# Aggregation Example
class Department:
    def __init__(self, name):
        self.name = name

class University:
    def __init__(self, name):
        self.name = name
        self.departments = []

    def add_department(self, dept):
        self.departments.append(dept)

uni = University("Tech University")
dept1 = Department("CS")
dept2 = Department("Math")

uni.add_department(dept1)
uni.add_department(dept2)

for d in uni.departments:
    print(d.name)
```

```python
# Composition Example
class Engine:
    def __init__(self):
        print("Engine created")

class Car:
    def __init__(self):
        self.engine = Engine()  # Composition: Engine cannot exist without Car
        print("Car created")

car = Car()
```
