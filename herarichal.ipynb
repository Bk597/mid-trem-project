```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.datasets import make_blobs
from sklearn.cluster import AgglomerativeClustering

from scipy.cluster.hierarchy import dendrogram, linkage
```

```python
X, y = make_blobs(
    n_samples=300,
    centers=4,
    cluster_std=1.2,
    random_state=42
)

print("Shape of dataset:", X.shape)
```

```python
plt.figure(figsize=(7,5))

plt.scatter(X[:,0], X[:,1])
plt.title("Unsupervised Dataset")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")

plt.show()
```

```python
plt.figure(figsize=(10,6))

linked = linkage(X, method='ward')

dendrogram(linked)

plt.title("Dendrogram")
plt.xlabel("Data Points")
plt.ylabel("Euclidean Distance")

plt.show()
```

```python
hc = AgglomerativeClustering(
    n_clusters=4,
    metric='euclidean',
    linkage='ward'
)

y_pred = hc.fit_predict(X)

print(y_pred[:10])
```

```python
plt.figure(figsize=(8,6))

plt.scatter(X[y_pred == 0, 0], X[y_pred == 0, 1], label='Cluster 1')
plt.scatter(X[y_pred == 1, 0], X[y_pred == 1, 1], label='Cluster 2')
plt.scatter(X[y_pred == 2, 0], X[y_pred == 2, 1], label='Cluster 3')
plt.scatter(X[y_pred == 3, 0], X[y_pred == 3, 1], label='Cluster 4')

plt.title("Hierarchical Clustering Results")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")

plt.legend()
plt.show()
```
