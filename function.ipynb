```python
# Global variable
x = 10

def show_global():
    print("Inside function, global x =", x)

show_global()
print("Outside function, global x =", x)
```

```python
# Global variable
x = 10

def modify_global():
    global x  # Tell Python we mean the global variable x
    x += 5
    print("Inside function, x =", x)

print("Before function, x =", x)
modify_global()
print("After function, x =", x)
```

```python
def show_local():
    # Local variable
    y = 5
    print("Inside function, local y =", y)

show_local()
#print(y)
```

```python
def outer():
    # Enclosed variable
    z = 20

    def inner():
        nonlocal z  # Use nonlocal to modify the enclosed variable
        z += 5
        print("Inside inner function, z =", z)

    inner()
    print("Inside outer function, z =", z)

outer()
```

```python
def process_data(data):
    
    def clean(x):
        return x.strip().lower()
    
    def is_valid(x):
        return len(x) > 3
    
    cleaned = list(map(clean, data))
    filtered = list(filter(is_valid, cleaned))
    
    return filtered

data = [" AI ", "ML", " Data ", "Py"]
print(process_data(data))
```

```python
def sum_list(nums):
    total = 0
    for num in nums:
        total += num
    return total

numbers = [1, 2, 3, 4, 5]
print("Sum is:", sum_list(numbers))

# Reusing the same function for another list
print("Sum is:", sum_list([10, 20, 30]))
```

```python
# Function to calculate grade
marks = float(input("Enter your marks: "))
def calculate_grade(mark):
    if mark >= 90:
        return 'A+'
    elif mark >= 80:
        return 'A'
    elif mark >= 70:
        return 'B'
    elif mark >= 60:
        return 'C'
    elif mark >= 50:
        return 'D'
    else:
        return 'F'

# Input from user

print("Your grade is:", calculate_grade(marks))
```
